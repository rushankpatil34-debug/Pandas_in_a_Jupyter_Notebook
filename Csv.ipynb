```python
# adding option based field
import numpy as np
conditions = [
    df["Marks"] >= 90,
    df["Marks"] >= 75,
    df["Marks"] >= 60,
    df["Marks"] >= 40
]
choices = ["A+", "A", "B", "C"]
df["Grade"] = np.select(conditions,choices,default="F")
print(df[["Name","Marks","Grade"]])
```
